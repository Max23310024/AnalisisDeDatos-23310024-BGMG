```python
# Ejercicio 1: Variacion porcentual
ventas_inicial = 200000
ventas_final = 260000

costos_inicial = 120000
costos_final = 150000

# a) Variacion porcentual de ventas
var_ventas = ((ventas_final - ventas_inicial) / ventas_inicial) * 100

# b) Variacion porcentual de costos
var_costos = ((costos_final - costos_inicial) / costos_inicial) * 100

# c) Utilidad inicial y final
utilidad_inicial = ventas_inicial - costos_inicial
utilidad_final = ventas_final - costos_final

# Variacion porcentual de utilidad
var_utilidad = ((utilidad_final - utilidad_inicial) / utilidad_inicial) * 100

print("--- Ejercicio 1: Variacion porcentual ---")
print(f"Variacion de ventas: {var_ventas:.2f}%")
print(f"Variacion de costos: {var_costos:.2f}%")
print(f"Utilidad inicial: ${utilidad_inicial:.2f}")
print(f"Utilidad final: ${utilidad_final:.2f}")
print(f"Variacion de utilidad: {var_utilidad:.2f}%")
```

```python
# Ejercicio 2: Promedio ponderado
hombres_total = 400
hombres_tasa = 0.35

mujeres_total = 600
mujeres_tasa = 0.45

hombres_aprueban = hombres_total * hombres_tasa
mujeres_aprueban = mujeres_total * mujeres_tasa

total_aprueban = hombres_aprueban + mujeres_aprueban
total_personas = hombres_total + mujeres_total

tasa_global = (total_aprueban / total_personas) * 100
promedio_simple = ((hombres_tasa + mujeres_tasa) / 2) * 100

print("--- Ejercicio 2: Promedio ponderado ---")
print(f"Hombres que aprueban: {hombres_aprueban:.0f}")
print(f"Mujeres que aprueban: {mujeres_aprueban:.0f}")
print(f"Tasa global (ponderada): {tasa_global:.2f}%")
print(f"Promedio simple (incorrecto): {promedio_simple:.2f}%")
```

```python
# Ejercicio 3: Paradoja de Simpson - Hospitales

# Hospital A

A_bajo_casos = 100

A_bajo_exitos = 98

A_alto_casos = 100

A_alto_exitos = 60

# Hospital B

B_bajo_casos = 300

B_bajo_exitos = 288

B_alto_casos = 100

B_alto_exitos = 55

A_bajo_tasa = (A_bajo_exitos / A_bajo_casos) * 100

A_alto_tasa = (A_alto_exitos / A_alto_casos) * 100

A_global = ((A_bajo_exitos + A_alto_exitos) / (A_bajo_casos + A_alto_casos)) * 100

B_bajo_tasa = (B_bajo_exitos / B_bajo_casos) * 100

B_alto_tasa = (B_alto_exitos / B_alto_casos) * 100

B_global = ((B_bajo_exitos + B_alto_exitos) / (B_bajo_casos + B_alto_casos)) * 100

print("--- Ejercicio 3: Paradoja de Simpson ---")

print(f"Hospital A - Bajo riesgo: {A_bajo_tasa:.2f}%")

print(f"Hospital A - Alto riesgo: {A_alto_tasa:.2f}%")

print(f"Hospital A - Global: {A_global:.2f}%")

print(f"Hospital B - Bajo riesgo: {B_bajo_tasa:.2f}%")

print(f"Hospital B - Alto riesgo: {B_alto_tasa:.2f}%")

print(f"Hospital B - Global: {B_global:.2f}%")
```

```python
# Ejercicio 4: Valor esperado - Poliza de seguro
precio_poliza = 50
prob_siniestro = 0.02
pago_siniestro = 2000

pago_esperado = prob_siniestro * pago_siniestro
ganancia_esperada = precio_poliza - pago_esperado

print("--- Ejercicio 4: Valor esperado ---")
print(f"Pago esperado por siniestro: ${pago_esperado:.2f}")
print(f"Ganancia esperada por poliza: ${ganancia_esperada:.2f}")
```

```python
# Ejercicio 5: Modelo lineal - Gasto vs Ingreso
gasto_fijo = 180
pendiente = 0.25

aumento_ingreso = 1000
aumento_gasto = pendiente * aumento_ingreso

ingreso_ejemplo = 2000
gasto_estimado = gasto_fijo + pendiente * ingreso_ejemplo

porcentaje_gasto = (gasto_estimado / ingreso_ejemplo) * 100

print("--- Ejercicio 5: Modelo lineal ---")
print(f"Gasto fijo mensual: ${gasto_fijo:.2f}")
print(f"Gasto marginal (por cada peso extra): ${pendiente:.2f}")
print(f"Si el ingreso sube ${aumento_ingreso:.2f}, el gasto sube: ${aumento_gasto:.2f}")
print(f"Con ingreso de ${ingreso_ejemplo:.2f}, gasto estimado: ${gasto_estimado:.2f}")
print(f"Ese gasto representa el {porcentaje_gasto:.2f}% del ingreso")
```
